```python
import pandas as pd

df = pd.read_csv("../macro_monthly_clean.csv", index_col=0, parse_dates=True)

df = df.reset_index().rename(columns={"index": "date"})

cols_needed = [
    "date",
    "treasury_1y",
    "treasury_2y",
    "treasury_5y",
    "treasury_10y"
]

hist_df = df[cols_needed].dropna()

hist_df.to_csv("../bonds_dashboards/bonds-dashboard/public/yield_history.csv", index=False)

print("Saved yield_history.csv → /public/yield_history.csv")
```
