```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_csv("../macro_monthly_clean.csv", index_col=0, parse_dates=True)

yield_cols = ["treasury_1y", "treasury_2y", "treasury_5y", "treasury_10y"]

df = df[yield_cols].dropna()
df.tail()
```

```python
latest = df.index.max()
curve = df.loc[latest, yield_cols]

plt.figure(figsize=(8,5))
plt.plot([1,2,5,10], curve.values, marker="o")
plt.title(f"US Treasury Yield Curve on {latest.date()}")
plt.xlabel("Maturity (years)")
plt.ylabel("Yield (%)")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(12,6))
for col in yield_cols:
    plt.plot(df.index, df[col], label=col)

plt.title("Treasury Yields Over Time (1Y, 2Y, 5Y, 10Y)")
plt.ylabel("Yield (%)")
plt.grid(True)
plt.legend()
plt.show()
```

```python
dates = ["2019-12-31", "2021-12-31", "2023-12-31"]

plt.figure(figsize=(8,5))
for d in dates:
    curve = df.loc[d, yield_cols]
    plt.plot([1,2,5,10], curve.values, marker="o", label=str(d))

plt.title("Yield Curve Comparison Across Years")
plt.xlabel("Maturity (years)")
plt.ylabel("Yield (%)")
plt.grid(True)
plt.legend()
plt.show()
```
